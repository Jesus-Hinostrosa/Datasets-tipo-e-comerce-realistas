# Dataset Sintético de E-commerce con Ruido

Este notebook genera un dataset realista con errores para practicar data cleaning.

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)
```

## Configuración

```python
n_customers = 300
n_orders = 2500
start_date = datetime(2023, 1, 1)
```

## Generación de Clientes

```python
customers = pd.DataFrame({
    "customer_id": range(1, n_customers + 1),
    "country": np.random.choice(["Mexico", "USA", "Colombia"], n_customers),
    "signup_date": [start_date + timedelta(days=int(x)) for x in np.random.randint(0, 365, n_customers)]
})

# Introducir errores
customers.loc[np.random.choice(customers.index, 10), "country"] = None
customers.loc[np.random.choice(customers.index, 5), "country"] = "Mx"

customers.head()
```

## Generación de Productos

```python
products = pd.DataFrame({
    "product_id": range(1, 21),
    "category": np.random.choice(["Electronics", "Clothing", "Home", "Beauty"], 20),
    "price": np.round(np.random.uniform(10, 500, 20), 2)
})

# Errores
products.loc[np.random.choice(products.index, 3), "category"] = "electrnics"
products.loc[np.random.choice(products.index, 2), "price"] = -50

products.head()
```

## Generación de Órdenes

```python
orders = []

for i in range(1, n_orders + 1):
    customer = random.choice(customers["customer_id"].values)
    product = random.choice(products["product_id"].values)
    
    order_date = start_date + timedelta(days=random.randint(0, 365))
    quantity = random.randint(1, 3)
    price = products.loc[products["product_id"] == product, "price"].values[0]
    total = quantity * price
    
    orders.append([
        i,
        customer,
        product,
        order_date,
        quantity,
        price,
        round(total, 2)
    ])

orders = pd.DataFrame(orders, columns=[
    "order_id",
    "customer_id",
    "product_id",
    "order_date",
    "quantity",
    "price",
    "total_amount"
])

orders.head()
```

## Introducción de Ruido

```python
# Nulos
orders.loc[np.random.choice(orders.index, 50), "total_amount"] = None

# Fechas futuras
orders.loc[np.random.choice(orders.index, 20), "order_date"] = datetime(2026, 1, 1)

# Duplicados
orders = pd.concat([orders, orders.sample(30)])

# Outliers
orders.loc[np.random.choice(orders.index, 10), "total_amount"] = orders["total_amount"] * 50

# Totales incorrectos
orders.loc[np.random.choice(orders.index, 20), "total_amount"] = 1

orders.head()
```

## Guardar archivos

```python
customers.to_csv("customers_dirty.csv", index=False)
products.to_csv("products_dirty.csv", index=False)
orders.to_csv("orders_dirty.csv", index=False)

print("Dataset generado correctamente 🚀")
```
